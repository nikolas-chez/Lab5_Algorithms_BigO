```python
# TASK 12: LINEAR SEARCH

def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# TASK 13: BINARY SEARCH

def binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# TASK 15: BUBBLE SORT

def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a) - 1):
        swapped = False

        for j in range(len(a) - i - 1):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return a, comparisons, swaps


data = [
    (ID_SUM + 3) % 100,
    (SEED_DIGIT + 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# TASK 16: SELECTION SORT

def selection_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a) - 1):
        minimum = i

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1

    return a, comparisons, swaps


sorted_data, comparisons, swaps = selection_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# TASK 17: INSERTION SORT

def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    shifts = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] <= key:
                break

            a[j + 1] = a[j]
            shifts += 1
            j -= 1

        a[j + 1] = key

    return a, comparisons, shifts


sorted_data, comparisons, shifts = insertion_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# TASK 18: EXECUTION-TIME MEASUREMENT

import time

data_size = 100 + ID_SUM * 10
performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

trials = 5

def measure_time(sort_function, data, trials=5):
    total_time = 0

    for _ in range(trials):
        start = time.perf_counter()
        sort_function(data)
        end = time.perf_counter()
        total_time += end - start

    return total_time / trials

for name, sort_function in [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort)
]:
    average_time = measure_time(sort_function, performance_data, trials)
    print(f"{name}: {average_time:.6f} seconds")
```

```python
# Exercise 1: Emergency Dispatch Record Search

LAST_NAME = "Nicolas"
STUDENT_ID = "TUPM-26-7931"
FAVORITE_ARTIST = "NIKI"

SEED_DIGIT = 1
ID_SUM = 28
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

ARTIST_SUM = sum(
    ord(letter.upper()) - ord("A") + 1
    for letter in FAVORITE_ARTIST
    if letter.isalpha()
)

# Generate the original incident records
incident_records = sorted(set(
    1000 + i * (SEED_DIGIT + 2) + ID_SUM
    for i in range(20)
))

# Rotate the records using SEED_DIGIT
rotation_point = SEED_DIGIT % len(incident_records)

rotated_records = (
    incident_records[rotation_point:]
    + incident_records[:rotation_point]
)

print("Original incident records:", incident_records)
print("Rotation point:", rotation_point)
print("Rotated incident records:", rotated_records)
print("Target values: ID_SUM =", ID_SUM,
      "| ARTIST_SUM =", ARTIST_SUM)
```

```python
# Select the target incident using student-specific values

target_index = (ID_SUM + ARTIST_SUM) % len(incident_records)
target_incident = incident_records[target_index]

print("Target index:", target_index)
print("Target incident ID:", target_incident)
```

```python
# Rotated Binary Search

def rotated_binary_search(records, target):
    left = 0
    right = len(records) - 1
    comparisons = 0

    while left <= right:
        mid = (left + right) // 2
        comparisons += 1

        if records[mid] == target:
            return mid, comparisons

        # Check whether the left half is sorted
        if records[left] <= records[mid]:
            comparisons += 1

            if records[left] <= target < records[mid]:
                right = mid - 1
            else:
                left = mid + 1

        # Otherwise, the right half is sorted
        else:
            comparisons += 1

            if records[mid] < target <= records[right]:
                left = mid + 1
            else:
                right = mid - 1

    return -1, comparisons


# Search for the target incident
binary_position, binary_comparisons = rotated_binary_search(
    rotated_records, target_incident
)

print("Target incident:", target_incident)
print("Position (0-based):", binary_position)
print("Position (1-based):", binary_position + 1 if binary_position != -1 else "Not found")
print("Binary Search comparisons:", binary_comparisons)
```

```python
# Linear Search

def linear_search(records, target):
    comparisons = 0

    for index, value in enumerate(records):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons


# Search for the same target using Linear Search
linear_position, linear_comparisons = linear_search(
    rotated_records, target_incident
)

print("Target incident:", target_incident)
print("Position (0-based):", linear_position)
print(
    "Position (1-based):",
    linear_position + 1 if linear_position != -1 else "Not found"
)
print("Linear Search comparisons:", linear_comparisons)
```

```python
# Compare execution time and results

from time import perf_counter

start = perf_counter()
binary_position, binary_comparisons = rotated_binary_search(
    rotated_records, target_incident
)
binary_time = perf_counter() - start

start = perf_counter()
linear_position, linear_comparisons = linear_search(
    rotated_records, target_incident
)
linear_time = perf_counter() - start

print("=== SEARCH PERFORMANCE COMPARISON ===")
print(f"Target incident: {target_incident}")
print(f"Binary Search position: {binary_position + 1 if binary_position != -1 else 'Not found'}")
print(f"Binary Search comparisons: {binary_comparisons}")
print(f"Binary Search time: {binary_time:.8f} seconds")
print()
print(f"Linear Search position: {linear_position + 1 if linear_position != -1 else 'Not found'}")
print(f"Linear Search comparisons: {linear_comparisons}")
print(f"Linear Search time: {linear_time:.8f} seconds")
print()
print("Time complexity:")
print("Rotated Binary Search: O(log n), for distinct values")
print("Linear Search: O(n)")
print("Auxiliary space complexity: O(1) for both algorithms")
print("Target found by both algorithms:",
      binary_position != -1 and linear_position != -1)
```

```python
# Exercise 2: University Enrollment Records

# Dataset A: Department A
dataset_a = [
    10000 + SEED_DIGIT * 100 + ID_SUM + i * (SEED_DIGIT + 3)
    for i in range(10)
]

# Dataset B: Department B
dataset_b = [
    20000 + ARTIST_LENGTH * 100 + NAME_LENGTH + i * (ARTIST_LENGTH + 2)
    for i in range(8)
]

print("Dataset A:", dataset_a)
print("Dataset B:", dataset_b)
print("Dataset A is sorted:", dataset_a == sorted(dataset_a))
print("Dataset B is sorted:", dataset_b == sorted(dataset_b))
print("Total records before merge:", len(dataset_a) + len(dataset_b))
```

```python
# Merge the two sorted datasets

def merge_sorted(dataset_a, dataset_b):
    i = 0
    j = 0
    comparisons = 0
    merged_dataset = []

    while i < len(dataset_a) and j < len(dataset_b):
        comparisons += 1

        if dataset_a[i] <= dataset_b[j]:
            merged_dataset.append(dataset_a[i])
            i += 1
        else:
            merged_dataset.append(dataset_b[j])
            j += 1

    # Add any remaining records
    merged_dataset.extend(dataset_a[i:])
    merged_dataset.extend(dataset_b[j:])

    return merged_dataset, comparisons


consolidated_dataset, merge_comparisons = merge_sorted(
    dataset_a, dataset_b
)

print("Consolidated dataset:", consolidated_dataset)
print("Merge comparisons:", merge_comparisons)
```

```python
# Verify the merged dataset

is_sorted = all(
    consolidated_dataset[i] <= consolidated_dataset[i + 1]
    for i in range(len(consolidated_dataset) - 1)
)

records_before = len(dataset_a) + len(dataset_b)
records_after = len(consolidated_dataset)

all_records_preserved = (
    records_before == records_after
    and sorted(consolidated_dataset) == sorted(dataset_a + dataset_b)
)

print("Records before merge:", records_before)
print("Records after merge:", records_after)
print("Ascending order verified:", is_sorted)
print("All records preserved:", all_records_preserved)
print("Merge comparisons:", merge_comparisons)
print("Time complexity: O(n + m)")
print("Auxiliary space complexity: O(n + m)")
```

```python
# Exercise 3: E-Commerce Order Processing

import random

# Generate student-specific order IDs
random.seed(SEED_DIGIT * 10000 + ID_SUM * 100 + ARTIST_SUM)

order_ids = [
    100000 + (
        i * (ID_SUM + ARTIST_SUM + 1)
        + random.randint(0, 999)
    ) % 900000
    for i in range(20)
]

# Ensure all order IDs are unique
order_ids = list(dict.fromkeys(order_ids))

while len(order_ids) < 20:
    new_id = 100000 + random.randint(0, 899999)
    if new_id not in order_ids:
        order_ids.append(new_id)

# Create the three input conditions
sorted_orders = sorted(order_ids)
reverse_orders = sorted(order_ids, reverse=True)
random_orders = random.sample(order_ids, len(order_ids))

print("Generated order IDs:", order_ids)
print("\nSorted input:", sorted_orders)
print("\nReverse-ordered input:", reverse_orders)
print("\nRandom input:", random_orders)
print("\nNumber of unique orders:", len(set(order_ids)))
```

```python
# Bubble Sort

def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for end in range(len(arr) - 1, 0, -1):
        swapped = False

        for j in range(end):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements


# Test Bubble Sort on all three input conditions
for condition, data in [
    ("Sorted", sorted_orders),
    ("Reverse-ordered", reverse_orders),
    ("Random", random_orders)
]:
    result, comparisons, movements = bubble_sort(data)

    print(f"\n{condition} input:")
    print("Sorted correctly:", result == sorted(data))
    print("Comparisons:", comparisons)
    print("Movements:", movements)
```

```python
# Selection Sort

def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 3

    return arr, comparisons, movements


# Test Selection Sort on all three input conditions
for condition, data in [
    ("Sorted", sorted_orders),
    ("Reverse-ordered", reverse_orders),
    ("Random", random_orders)
]:
    result, comparisons, movements = selection_sort(data)

    print(f"\n{condition} input:")
    print("Sorted correctly:", result == sorted(data))
    print("Comparisons:", comparisons)
    print("Movements:", movements)
```

```python
# Insertion Sort

def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements


# Test Insertion Sort on all three input conditions
for condition, data in [
    ("Sorted", sorted_orders),
    ("Reverse-ordered", reverse_orders),
    ("Random", random_orders)
]:
    result, comparisons, movements = insertion_sort(data)

    print(f"\n{condition} input:")
    print("Sorted correctly:", result == sorted(data))
    print("Comparisons:", comparisons)
    print("Movements:", movements)
```

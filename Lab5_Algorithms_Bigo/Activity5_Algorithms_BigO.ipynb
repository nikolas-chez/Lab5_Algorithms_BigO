```python

# TASK 8: LINEAR GROWTH

def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# TASK 9: BIG O CLASSIFICATION

def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n²)={n*n:7d}"
    )
```

```python
# TASK 10: SPACE COMPLEXITY

def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# TASK 11: BEST, AVERAGE, AND WORST CASE

def search_case(data, target):
    for i, value in enumerate(data, 1):
        if value == target:
            return i
    return len(data)

data = [11, 22, 33, 44, 55]

for case, target in {
    "Best": 11,
    "Average": 33,
    "Worst": 99
}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
# TASK 12: LINEAR SEARCH

def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# TASK 13: BINARY SEARCH

def binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# TASK 14: SEARCH COMPARISON

data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

linear_index, linear_comparisons = linear_search(data, target)
binary_index, binary_comparisons = binary_search(data, target)

print("Target:", target)
print("Linear Search Index:", linear_index)
print("Linear Search Comparisons:", linear_comparisons)
print("Binary Search Index:", binary_index)
print("Binary Search Comparisons:", binary_comparisons)
print("Comparison Reduction:", linear_comparisons - binary_comparisons)
```

```python
# TASK 15: BUBBLE SORT

def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a) - 1):
        swapped = False

        for j in range(len(a) - i - 1):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return a, comparisons, swaps


data = [
    (ID_SUM + 3) % 100,
    (SEED_DIGIT + 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# TASK 16: SELECTION SORT

def selection_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0

    for i in range(len(a) - 1):
        minimum = i

        for j in range(i + 1, len(a)):
            comparisons += 1

            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1

    return a, comparisons, swaps


sorted_data, comparisons, swaps = selection_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# TASK 17: INSERTION SORT

def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    shifts = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if a[j] <= key:
                break

            a[j + 1] = a[j]
            shifts += 1
            j -= 1

        a[j + 1] = key

    return a, comparisons, shifts


sorted_data, comparisons, shifts = insertion_sort(data)

print("Original Data:", data)
print("Sorted Data:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# TASK 18: EXECUTION-TIME MEASUREMENT

import time

data_size = 100 + ID_SUM * 10
performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

trials = 5

def measure_time(sort_function, data, trials=5):
    total_time = 0

    for _ in range(trials):
        start = time.perf_counter()
        sort_function(data)
        end = time.perf_counter()
        total_time += end - start

    return total_time / trials

for name, sort_function in [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort)
]:
    average_time = measure_time(sort_function, performance_data, trials)
    print(f"{name}: {average_time:.6f} seconds")
```

```python
# TASK 19: SORTING PERFORMANCE COMPARISON

algorithms = [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort)
]

print("SORTING PERFORMANCE COMPARISON")
print("-" * 65)

for name, sort_function in algorithms:
    start = time.perf_counter()
    sorted_data, comparisons, movements = sort_function(performance_data)
    elapsed = time.perf_counter() - start

    print(f"Algorithm: {name}")
    print(f"Comparisons: {comparisons}")
    print(f"Movements: {movements}")
    print(f"Execution Time: {elapsed:.6f} seconds")
    print(f"Sorted Correctly: {sorted_data == sorted(performance_data)}")
    print("-" * 65)
```
